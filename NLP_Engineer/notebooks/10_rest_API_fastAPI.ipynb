The goal of Week 10 is to expose the NLP capabilities developed throughout the project through a production-oriented REST API using FASTAPI.

Application Programming Interface (API) is a way for one program to communicate with another program. 

Representational State Transfer (REST) API is an API that uses standard HTTP methods and URLs to interact with resources or services. The browser sends an HTTP request. Then the server sends an HTTP response. A REST API uses that same basic communication mechanism. The major HTTP methods are GET, POST, PUT, PATCH, and DELETE. 

REST is an architectural approach while FastAPI is a Python framework. FastAPI gives us tools that make it much easier to build the REST API.

**What is response caching?**
- If the API receives the exact same request again, don't perform all the NLP/database/model work again. Just return the result you already calculated.
- So, if the same request comes in repeatedly, caching prevents unnecessary processing. 

To utilize response caching, I will use a dictionary to satisfy this requirement.

**What is rate limiting?**
- The requirement for this week's work says to have rate limiting (10 requests/second per IP), which means your API should prevent one client/IP address from sending an unlimited number of requests in a short period.
- By setting a rate limit, it protects your API from accidental request loops and clients sending too many requests.

For FastAPI applications, I will use slowapi, which should be sufficient enough for this project.

Currently without using a Docker, through the terminal, I run the API using uvicorn api.main:app --reload. But with Docker, when we start the container, the uvicorn runs inside the container to then result in the FastAPI to run. 

Also, without Docker, clients may have to install certain packages to get the API working, but with Docker, it can create the environment for them, so the API can be packaged and deployed as a self-contained application.

The requirements.txt file tells Docker which Python packages the API needs.

**More about the Docker Deployment**

This *docker build -t real-estate-nlp-api .* command in the home directory was ran in which builds a Docker image for the NLP API using the Dockerfile in the current directory, and give the image the name real-estate-nlp-api. 

Once the image is built, it can be run as a container using: *docker run -p 8000:8000 real-estate-nlp-api*. This maps port 8000 on the local machine to port 8000 inside the container, allowing the FastAPI application to be accessed through localhost:8000. So, it basically starts the container and then starts the API.
